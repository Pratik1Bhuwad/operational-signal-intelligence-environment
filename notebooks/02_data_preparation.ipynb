```python
import pandas as pd
```

```python
demand = pd.read_csv("D:/operational-signal-intelligence-environment/data/raw/national_demand.csv")

generation = pd.read_csv("D:/operational-signal-intelligence-environment/data/raw/generation_mix.csv")

forecast = pd.read_csv("D:/operational-signal-intelligence-environment/data/raw/forecast_performance.csv")
```

```python
demand['SETTLEMENT_DATE'] = pd.to_datetime(
    demand['SETTLEMENT_DATE'],
    format='%d-%m-%Y'
)

generation['DATETIME'] = pd.to_datetime(
    generation['DATETIME']
)

forecast['Datetime'] = pd.to_datetime(
    forecast['Datetime']
)
forecast['Date'] = pd.to_datetime(
    forecast['Date']
)
```
