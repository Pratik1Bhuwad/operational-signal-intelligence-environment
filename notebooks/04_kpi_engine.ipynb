```python
import pandas as pd
```

```python
master_df = pd.read_csv(
    "D:/operational-signal-intelligence-environment/data/processed/master_operational_dataset.csv",
    parse_dates=["Datetime"]
)
```

```python
avg_demand = master_df["ND"].mean()
```

```python
peak_demand = master_df["ND"].max()
```

```python
avg_generation = master_df["GENERATION"].mean()
```

```python
renewable_percentage = (
    master_df["RENEWABLE"].sum()
    /
    master_df["GENERATION"].sum()
) * 100
```

```python
fossil_percentage = (
    master_df["FOSSIL"].sum()
    /
    master_df["GENERATION"].sum()
) * 100
```

```python
avg_carbon_intensity = master_df["CARBON_INTENSITY"].mean()
```

```python
forecast_accuracy = 100 - master_df["APE"].mean()
```

```python
avg_forecast_error = master_df["Absolute_Error"].mean()
```

```python
peak_renewable_generation = master_df["RENEWABLE"].max()
```

```python
peak_carbon_intensity = master_df["CARBON_INTENSITY"].max()
```

```python
peak_forecast_error = master_df["Absolute_Error"].max()
```

```python
kpis = {
    "Average National Demand (MW)": avg_demand,
    "Peak National Demand (MW)": peak_demand,
    "Average Total Generation (MW)": avg_generation,
    "Renewable Generation Share (%)": renewable_percentage,
    "Fossil Generation Share (%)": fossil_percentage,
    "Average Carbon Intensity (gCO₂/kWh)": avg_carbon_intensity,   
    "Forecast Accuracy (%)": forecast_accuracy,
    "Average Forecast Error (MW)": avg_forecast_error,
    "Peak Renewable Generation (MW)": peak_renewable_generation,
    "Peak Carbon Intensity (gCO2/kWh)": peak_carbon_intensity,
    "Peak Forecast Error (MW)": peak_forecast_error

}
print("Key Performance Indicators (KPIs):",kpis)
```

```python
kpi_df = pd.DataFrame(
    list(kpis.items()),
    columns=["KPI", "Value"]
)
```
