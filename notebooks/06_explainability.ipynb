```python
import pandas as pd

master_df = pd.read_csv(
    "D:/operational-signal-intelligence-environment/data/processed/master_with_signals.csv",
    parse_dates=["Datetime"]
)
```

Investigate Demand Spikes

```python
demand_spikes = master_df[
    master_df["Demand_Spike"] == True
]
```

```python
print("Overall Average Demand:",
      master_df["ND"].mean())

print("Demand During Spikes:",
      demand_spikes["ND"].mean())
```

Investigate Forecast Failures

```python
forecast_failures = master_df[
    master_df["Forecast_Failure"] == True
]
```

```python
forecast_failures[
    [
        "Demand_Forecast",
        "Demand_Outturn",
        "Absolute_Error"
    ]
].describe()
```

Investigate High Carbon Periods

```python
high_carbon = master_df[
    master_df["High_Carbon_Period"] == True
]
```

```python
print("Average Fossil During High Carbon:",
      high_carbon["FOSSIL"].mean())

print("Overall Average Fossil:",
      master_df["FOSSIL"].mean())
```

```python
print("Average Renewable During High Carbon:",
      high_carbon["RENEWABLE"].mean())

print("Overall Average Renewable:",
      master_df["RENEWABLE"].mean())
```

Investigate Low Renewable Windows

```python
low_renewable = master_df[
    master_df["Low_Renewable_Window"] == True
]
```

```python
low_renewable[
    [
        "RENEWABLE",
        "FOSSIL",
        "CARBON_INTENSITY"
    ]
].describe()
```

Simple Explainability Table

```python
explanations = pd.DataFrame({
    "Signal": [
        "Demand Spike",
        "Forecast Failure",
        "High Carbon Period",
        "Low Renewable Window"
    ],
    "Evidence": [
        "Average demand during spikes is significantly higher than normal demand.",
        "Forecast failures show average error above 2500 MW.",
        "High carbon periods have much higher fossil generation and lower renewable generation.",
        "Low renewable periods are associated with elevated carbon intensity."
    ]
})

explanations
```

```python
explanations.to_csv(
    "D:/operational-signal-intelligence-environment/outputs/explainability_summary.csv",
    index=False
)
```

```python
import pandas as pd

comparison_df = pd.DataFrame({
    "Metric": [
        "Overall Average Demand",
        "Demand During Spikes",
        "Overall Average Fossil",
        "Fossil During High Carbon",
        "Overall Average Renewable",
        "Renewable During High Carbon"
    ],
    "Value": [
        master_df["ND"].mean(),
        demand_spikes["ND"].mean(),
        master_df["FOSSIL"].mean(),
        high_carbon["FOSSIL"].mean(),
        master_df["RENEWABLE"].mean(),
        high_carbon["RENEWABLE"].mean()
    ]
})

comparison_df
```
