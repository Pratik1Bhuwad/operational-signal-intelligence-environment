```python
import pandas as pd

demand = pd.read_csv(
    "D:/operational-signal-intelligence-environment/data/processed/demand.csv",
    parse_dates=["Datetime"]
)

generation = pd.read_csv(
    "D:/operational-signal-intelligence-environment/data/processed/generation.csv",
    parse_dates=["DATETIME"]
)

forecast = pd.read_csv(
    "D:/operational-signal-intelligence-environment/data/processed/forecast.csv",
    parse_dates=["Datetime"]
)
```

```python
print(demand["Datetime"].min())
print(demand["Datetime"].max())

print(generation["DATETIME"].min())
print(generation["DATETIME"].max())

print(forecast["Datetime"].min())
print(forecast["Datetime"].max())
```

```python
demand_keep = demand[
    [
        "Datetime",
        "ND",
        "TSD",
        "ENGLAND_WALES_DEMAND",
        "EMBEDDED_WIND_GENERATION",
        "EMBEDDED_SOLAR_GENERATION"
    ]
]
```

```python
generation_keep = generation[
    [
        "DATETIME",
        "GENERATION",
        "RENEWABLE",
        "FOSSIL",
        "CARBON_INTENSITY"
    ]
]
```

```python
generation_keep = generation_keep.rename(
    columns={"DATETIME": "Datetime"}
)
```

```python
forecast_keep = forecast[
    [
        "Datetime",
        "Demand_Forecast",
        "Demand_Outturn",
        "Absolute_Error",
        "APE"
    ]
]
```

```python
master_df = demand_keep.merge(
    generation_keep,
    on="Datetime",
    how="inner"
)
```

```python
master_df.shape
```

```python
master_df = master_df.merge(
    forecast_keep,
    on="Datetime",
    how="inner"
)
```
